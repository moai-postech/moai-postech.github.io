# 1. First Result, First Test

AI가 만든 코드를 받았다고 가정하고, 명세와 실행 결과를 직접 대조합니다. 외부 API 호출은 없습니다.

먼저 자신이 가진 AI 도구에 “score가 threshold 이상이면 1을 반환하는 함수를 작성하고 경계값 테스트를 넣어 달라”고 요청해 보세요. 도구가 없으면 아래 후보 함수를 검토하면 됩니다.

```python
def candidate(score, threshold=0.5):
    return int(score > threshold)

cases = [(0.2, 0), (0.5, 1), (0.8, 1)]
print([(x, candidate(x), expected) for x, expected in cases])
assert candidate(0.5) != 1
```

## 과제

- 명세를 만족하도록 비교 연산자를 고치세요.
- 음수 입력과 빈 입력의 처리 방식을 명세하고, 테스트를 추가하세요.

## 해답

```python
def checked(score, threshold=0.5):
    if score is None:
        return None
    if not 0 <= score <= 1:
        raise ValueError("score must be in [0, 1]")
    return int(score >= threshold)

assert all(checked(x) == expected for x, expected in cases)
assert checked(None) is None
try:
    checked(-0.1)
except ValueError:
    print("Out-of-range input rejected")
else:
    raise AssertionError("negative input was not rejected")
print("Acceptance tests passed")
```
