# Clustering

같은 입력에 scale과 K를 바꾸어 clustering이 정의에 의존하는 것을 확인합니다.

핵심 셀은 Python 3 + NumPy만 사용합니다. 데이터 다운로드와 API 키가 필요하지 않습니다.

## 입력과 거리

```python
import numpy as np
X=np.array([[0.,0.],[0.,1.],[1.,.2],[4.,4.],[4.,5.],[5.,4.2]])
def kmeans(X,k,seed=2):
    rng=np.random.default_rng(seed); c=X[rng.choice(len(X),k,replace=False)].copy()
    for _ in range(50):
        lab=((X[:,None,:]-c[None,:,:])**2).sum(2).argmin(1)
        new=np.array([X[lab==j].mean(0) if np.any(lab==j) else c[j] for j in range(k)])
        if np.allclose(c,new): break
        c=new
    sse=((X-c[lab])**2).sum()
    return lab,c,sse
print(kmeans(X,2))
```

## K와 objective

```python
for k in [1,2,3]:
    print("K",k,"SSE",kmeans(X,k)[2])
```

## 단위와 neighbor

```python
points=np.array([[0.,0.],[1.,1.],[4.,0.],[4.,3.]])
query=np.array([1.,0.])
for scale in [.2,1.,5.]:
    dist=np.linalg.norm((points-query)*[1,scale],axis=1)
    print("scale",scale,"distances",dist,"nearest",np.flatnonzero(np.isclose(dist,dist.min())))
```

## 과제

- K=2에서 seed를 바꾸어 assignment가 유지되는지 확인하세요.
- 두 번째 축 scale이 0.2와 5일 때 query의 최근접 점을 구하세요.

## 해답과 검증

```python
for seed in range(4): print(seed,kmeans(X,2,seed)[2])
dsmall=np.linalg.norm((points-query)*[1,.2],axis=1)
dlarge=np.linalg.norm((points-query)*[1,5],axis=1)
assert np.argmin(dsmall)==1 and np.argmin(dlarge)==0
print("nearest switch verified")
```
