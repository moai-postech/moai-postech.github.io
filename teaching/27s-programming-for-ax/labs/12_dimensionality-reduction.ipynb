# Dimensionality Reduction

SVD로 PCA를 계산하고 train 기준 변환·복원 오차·scale 효과를 확인합니다.

핵심 셀은 Python 3 + NumPy만 사용합니다. 데이터 다운로드와 API 키가 필요하지 않습니다.

## train에서 basis 학습

```python
import numpy as np
X=np.array([[-2.,-1.],[-1.,-.6],[0.,0.],[1.,.8],[2.,1.1]])
mu=X.mean(0); Xc=X-mu
U,s,Vt=np.linalg.svd(Xc,full_matrices=False)
Z=Xc@Vt[:1].T
Xhat=Z@Vt[:1]+mu
print("variance ratio",s*s/(s*s).sum())
print("rank1 MSE",np.mean((X-Xhat)**2))
```

## 새 관측은 같은 basis로 변환

```python
new=np.array([[3.,1.7]])
score=(new-mu)@Vt[:1].T
print("new score",score)
assert score.shape==(1,1)
```

## 작은 분산과 중요한 label

```python
rng=np.random.default_rng(8)
y=np.repeat([0,1],20)
features=np.c_[rng.normal(0,4,40), y*.4+rng.normal(0,.02,40)]
_,_,axes=np.linalg.svd(features-features.mean(0),full_matrices=False)
print("first PC",axes[0])
print("class mean difference",features[y==1].mean(0)-features[y==0].mean(0))
```

## 해답과 검증

```python
full=(Xc@Vt.T)@Vt+mu
assert np.allclose(full,X)
_,_,scaled_axes=np.linalg.svd(Xc*[1,100],full_matrices=False)
print("original axis",Vt[0],"scaled axis",scaled_axes[0])
print("full reconstruction verified")
```
