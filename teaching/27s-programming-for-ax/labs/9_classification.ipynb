# Classification

확률과 threshold를 분리하고 실제 confusion matrix, precision, recall, 비용을 계산합니다.

핵심 셀은 Python 3 + NumPy만 사용합니다. 데이터 다운로드와 API 키가 필요하지 않습니다.

## 고정된 작은 입력

```python
import numpy as np
y=np.array([0,0,1,0,1,1])
p=np.array([.05,.25,.4,.55,.75,.9])
assert y.shape == p.shape
```

## 운영 지표

```python
def evaluate(t):
    a=p>=t
    tp=int(((y==1)&a).sum()); fp=int(((y==0)&a).sum())
    fn=int(((y==1)&~a).sum()); tn=int(((y==0)&~a).sum())
    precision=tp/(tp+fp) if tp+fp else 0.
    recall=tp/(tp+fn) if tp+fn else 0.
    return tp,fp,tn,fn,precision,recall
print(evaluate(.5))
```

## threshold와 비용

```python
thresholds=np.linspace(0,1,21)
costs=np.array([evaluate(t)[1]+5*evaluate(t)[3] for t in thresholds])
best=thresholds[costs.argmin()]
print("validation choice", best, "cost", costs.min())
```

## 확률 오차

```python
brier=np.mean((p-y)**2)
logloss=-np.mean(y*np.log(p)+(1-y)*np.log1p(-p))
print("Brier", brier, "logloss", logloss)
```

## 과제

- threshold 0.3과 0.8에서 precision과 recall을 비교하세요.
- positive label이 드문 1,000건에서 모두 negative인 예측의 accuracy와 recall을 계산하세요.

## 해답과 검증

```python
print("t=.3",evaluate(.3)); print("t=.8",evaluate(.8))
rare=np.r_[np.zeros(990,dtype=int),np.ones(10,dtype=int)]
print("accuracy", np.mean(rare==0), "recall", 0.)
assert evaluate(.5)[:4]==(2,1,2,1)
assert np.isclose(np.mean(rare==0), .99)
```
