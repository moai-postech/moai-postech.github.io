# Deep Learning

XOR의 표현과 gradient를 각각 실행해 표현 가능성과 학습 계산을 구분합니다.

핵심 셀은 Python 3 + NumPy만 사용합니다. 데이터 다운로드와 API 키가 필요하지 않습니다.

## XOR forward

```python
import numpy as np
X=np.array([[0.,0.],[0.,1.],[1.,0.],[1.,1.]])
h1=np.maximum(0,X.sum(1)); h2=np.maximum(0,X.sum(1)-1)
output=h1-2*h2
print(output)
assert np.array_equal(output,[0,1,1,0])
```

## 수치 미분과 분석 미분

```python
def loss(w): return .5*(2*w-3)**2
w=1.; eps=1e-5
num=(loss(w+eps)-loss(w-eps))/(2*eps)
grad=2*(2*w-3)
print(num,grad)
assert np.isclose(num,grad)
```

## 완전한 scalar 학습 루프

```python
w=0.; trace=[]
for step in range(30):
    prediction=2*w
    L=.5*(prediction-3)**2
    grad=2*(prediction-3)
    w-=.1*grad
    trace.append(L)
print("weight",w,"prediction",2*w,"loss",trace[-1])
assert trace[-1]<trace[0]
```

```python
bad=h1
print("without second hidden unit",bad)
assert bad[-1]!=output[-1]
w=0.; divergent=[]
for _ in range(10):
    divergent.append(loss(w)); w-=.6*2*(2*w-3)
print("large-rate losses",divergent)
assert divergent[-1]>divergent[0]
```
