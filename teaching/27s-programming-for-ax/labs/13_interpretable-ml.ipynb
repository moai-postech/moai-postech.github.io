# Interpretable ML

작은 알려진 모델로 permutation, ICE, exact Shapley contribution을 검증합니다.

핵심 셀은 Python 3 + NumPy만 사용합니다. 데이터 다운로드와 API 키가 필요하지 않습니다.

## 알려진 모델

```python
import numpy as np
def f(a,b): return 2*np.asarray(a)+np.asarray(b)+.5*np.asarray(a)*np.asarray(b)
a=np.array([0.,1.,2.]); b=np.array([0.,2.,4.])
y=f(a,b)
print("outputs",y)
```

## 고정 모델의 permutation

```python
baseline=np.mean((y-f(a,b))**2)
shuffled=np.mean((y-f(a[::-1],b))**2)
print("baseline loss",baseline,"shuffle loss",shuffled)
```

## ICE와 평균

```python
grid=np.linspace(0,3,5)
ice=np.array([f(grid,bi) for bi in [0.,2.,4.]])
print("ICE rows",ice)
print("PDP",ice.mean(0))
```

## 두 feature exact attribution

```python
a0,b0=2.,4.
pa=.5*((f(a0,0)-f(0,0))+(f(a0,b0)-f(0,b0)))
pb=.5*((f(0,b0)-f(0,0))+(f(a0,b0)-f(a0,0)))
print("contributions",pa,pb)
assert np.isclose(pa+pb,f(a0,b0)-f(0,0))
```

## 과제

- baseline을 (1,1)로 바꾸어 두 feature attribution을 다시 계산하세요.
- prediction 10 이상인 현재점 (2,4)에서 a만 줄여 decision을 바꾸세요.

## 해답과 검증

```python
a_ref,b_ref=1.,1.
qa=.5*((f(a0,b_ref)-f(a_ref,b_ref))+(f(a0,b0)-f(a_ref,b0)))
qb=.5*((f(a_ref,b0)-f(a_ref,b_ref))+(f(a0,b0)-f(a0,b_ref)))
assert np.isclose(qa+qb,f(a0,b0)-f(a_ref,b_ref))
print("new reference",qa,qb)
assert f(1.,4.)<10<=f(2.,4.)
print("counterfactual",f(1.,4.))
```
